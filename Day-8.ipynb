```python
#inheritance-->Inheriting one class to another class
class Animal:#parent class
    def eat(self):
        print("animal eats")
class Lion(Animal):#Child class
    def king(self):
        print("king of the jungle")
l=Lion()
l.eat()
l.king()
```

```python
#Single inheritance
#Multi level Inheritance
#Grand parent-->parent-->child
class Animal:
    def eat(self):
        print("animal eats")
class Lion(Animal):#Child class
    def king(self):
        print("king of the jungle")
class Cub(Lion):
    def sleep(self):
        print("cub sleeps")
c=Cub()
c.sleep()
c.king()
c.eat()
```

```python
#Hierrarchical Inheritance_>one parent,multiple child
class Animal:
    def eat(self):
        print("animal eats")
class Lion(Animal):
    def king(self):
        print("king of the jungle")
class Dog(Animal):
    def guard(self):
        print("guards the home")
d=Dog()
d.eat()
d.guard()
l=Lion()
l.eat()
l.king()
```

```python
#Multiple Inheritance->one child class with multiple parent class
class Animal:
    def eat(self):
        print("animal eats")
class WildAnimal:
    def hunt(self):
        print("Animal hunts")
class Lion(Animal,WildAnimal):
    def king(self):
        print("king of the jungle")
l=Lion()
l.king()
l.hunt()
l.eat()
```

```python
#Super keyword-->used to access point method
class Animal:
    def eat(self):
        print("animal eats")
class Lion(Animal):
    def king(self):
        super().eat()
        print("king of the jungle")
l=Lion()
l.king()
```

```python
#Abstraction
from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod#declared but not defined
    def start(self):
        pass
class Car(Vehicle):
    def start(self):
        print("Car is starting")
c=Car()
c.start()
```
