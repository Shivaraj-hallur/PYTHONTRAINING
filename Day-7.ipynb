```python
#Functions or Methods
#Parametrized function
name=input("enter the name")
def greet(name):
    print("Welcome"+name)
greet("Alice")
greet(name)
```

```python
str=input("enter the name")
def greet(name):
    print("Welcome"+name)
greet("Alice")
greet(str)
```

```python
#add two number using method
a=input(input("enter a"))
b=input(input("enter b"))
def add(a,b):
    print(a+b)
add(a,b)
add(5,3)
add(12,4)
```

```python
#default value to the parameter
def greet(name="guest"):
    print("welcome "+name)
greet("Alice")
greet()
```

```python
#positional Arguments
def student(name,age):
    print("name",name)
    print("age",age)
student(age=20,name="alice")
```

```python
#a^2+b^2+2ab
a=2
b=3
def asquare(a):
    return a*a
def bsquare(b):
    return b*b
def twoab(a,b):
    return 2*a*b
x=asquare(a)
y=asquare(b)
z=twoab(a,b)
print(x+y+z)
```

```python
#returning multiple values
a=5
b=3
def operation(a,b):
    return a+b,a-b
sum,diff=operation(a,b)#tuple unpacking
print(sum,diff)
```

```python
#arbitrary arguments or variable length arguments
def add(*nums):
    print(nums)
add(1,2,3,4)
```

```python
#lambda expression
#method_name=lambda parameters:statements
add=lambda a,b:a+b#return value by default
print(add(2,3))
```

```python
#check even or odd using lambda
check=lambda n:n%2==0
print(check(4))
```

```python
#Recursion->calling the same method again and again
#base case-->stop the recursion
#recursion case->start the recursion
def sample(n):
    if n>10:
        return
    print(n)#1,2
    sample(n+1)
sample(1)
```

```python
#Factorial form using recursion
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)

fact(4)
#4*fact(3)
#3*fact(2)
#2*fact(1)
#return 1
```
