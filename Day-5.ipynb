```python
#Jumping statements(break,continue,pass)
#break-->used to break a loop
for i in range(0,5):
    if i==3:
        break
    print(i)
```

```python
#continue-->used to skip a iteration
for i in range(0,5):
    if i==2 or i==3:
        continue
    print(i)
```

```python
#pass-->do nothing(place holder)
for i in range(0,5):
    pass
```

```python
#Find the first non repeating character in a string
#Input-->"swiss"
#ouput-->'w'

str="swiss"
for ch in str:
    if str.count(ch)==1:
        print(ch)
        break
else:
    print("no one repeated character")
```

```python
#creating a list
nums=[1,2,3,4]#Integer lists
nums1=["java","python"]#string list
nums2=[1,"java",20.5]#all data types
nums3=[]#empty list
print('Tiger')
```

```python
#Accessing the list
nums=[1,2,3,4]
print(nums[1])
print(nums[-1])
print(nums[0:3])
print(nums[0:4:2])
```

```python
#Update the existing element
nums=[1,2,3,4]
nums[0]=50
print(nums)
```

```python
#add the new element
nums=[1,2,3,4]
nums.append(5)#add element in last
nums.insert(1,6)
print(nums)
```

```python
#Deleting the element
#remove->deletion based on element
nums=[1,2,3,4]
nums.remove(2)
print(nums)
```

```python
#pop-->deletion based on index
nums=[1,2,3,4]
nums.pop(2)
print(nums)
```

```python
#clear-->all elements will be deleted
nums=[1,2,3,4]
nums.pop(2)
nums.clear()
print(nums)
```

```python
#Aggegrate functions
nums=[1,2,3,4,5]
print(len(nums))#length of the list
print(sum(nums))#sum of the list
print(min(nums))#minimum element in the list
print(max(nums))#maximum element in the list
print(nums.count(1))#return the count of the element
```

```python
nums=[2,4,5,6,1]
nums.sort()
nums.reverse()
print(nums.index(6))
print(nums)
```

```python
#traverse the list
nums=[1,2,3,4,5]
for num in nums:
    print(num)
```

```python
nums=[1,2,3,4,5]#travesing based on element
for num in nums:
    print(num,end=" ")
```

```python
nums=[1,2,3,4,5]
for i in range(0,len(nums)):#travesing based on index
    print(nums[i],end=" ")
```
