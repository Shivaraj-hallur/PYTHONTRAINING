```python
#sieve of eratosthenes
n=50
prime=[True]*(n+1)
i=2
while i*i<=n:
    if prime[i]:
        for j in range(i*i,n+1,i):
            prime[j]=False
        i=i+1
    for i in range(2,n+1):
        if prime[i]:
            print(i,end=" ")
```
