```python
#find the factorial of a number
#n!=1*2..n
#1!=1
#2!=1*2=2
#3!=1*2*3=6
#4!=1*2*3*4=24
#5!=1*2*3*4*4=120

#n=5
#variable as fact=1
#loop runs 1 to n+1
#i=1 fact=fact*i--->1(1*1)
#i=2 fact=fact*i--->2(1*2)
#i=3 fact=fact*i--->6(2*3)
#i=4 fact=fact*i--->24(6*4)
#i=5 fact=fact*i--->120(24*5)
```

```python
n=2
fact=1
for i in range(1,n+1):
    fact=fact*i
print(fact)
```

```python
#reverse an integer
#Input-->456
#output-->654

num=456
rev=0
while num>0:
    rem=num%10
    rev=(rev*10)+rem
    num=num//10
print(rev)
```

```python
#Pattern programming
n=4
for i in range(1,n+1):
    for j in range(i):
        print('*',end="")
    print()
```

```python
n=4
for i in range(n,0,-1):
    for j in range(i):
        print('*',end="")
    print()
```
