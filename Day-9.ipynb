```python
#EXCEPTION-->Abnormal condition with interrupts the flow of the program
a=4
b=2
print("start")#start
print(a/b)#exception
print("end")
```

```python
#Exception Handling-->handle the exception and complete the flow of execution
#Keywords used in exception handling
#1 try
#2 except
#3 finally
#4 raise
```

```python
#try-->Whichever line of code may throw the exception 
a=4
b=0
print("Start")
try:
    print(a/b)
except ZeroDivisionError:
    print("exception caught")
print("end")
```

```python
#Catching multiple exception
nums=[1,2,3,4]
try:
    nums[7]=4/0
except (ZeroDivisionError,IndexError):
    print("exception caught")
print("end")
```

```python
#generic exception handling
nums=[1,2,3,4]
try:
    nums[7]=4/2
except Exception:
    print("exception caught")
print("end")

```

```python
#printing the message
nums=[1,2,3,4]
try:
    nums[7]=4/2
except Exception as e:
    print(e)
print("end")
```

```python
#finally-->always execute(even try/Except fails)
nums=[1,2,3,4]
try:
    nums[7]=4/0
except IndexError:
    print("Exception caught")
finally:
    print("end")
```

```python
#raise->create custom exception
class InvalidAgeError(Exception):
    pass
age=int(input("Enter age"))
if age<18:
    raise InvalidAgeError("Age must be greater than 18")
else:
    print("eligible")
```

```python
#balance,withdraw
#withdraw>balance-->raise("Insufficient balance")
#or print("amount dispensed")
```

```python
balance = 5000

withdraw = int(input("Enter withdraw amount: "))

if withdraw > balance:
    raise Exception("Insufficient balance")
else:
    print("Amount dispensed")
    balance = balance - withdraw
    print("Remaining balance:", balance)
```
